### Modifying Blocks: Adding 'water' block

### Adjusting World Generation: Adding water and varying terrain

```python
# Now, let's modify the 'generate_imaland' function.
# We'll replace the old function with a new one that includes water and more varied terrain.

old_generate_func_start = game_code.find("def generate_imaland():")
old_generate_func_end = game_code.find("# --- Game Loop (Survival Logic) ---", old_generate_func_start)

if old_generate_func_start != -1 and old_generate_func_end != -1:
    new_generate_imaland_code = '''def generate_imaland():
    for z in range(50):
        for x in range(50):
            # Introduce more complex terrain with varying heights and water
            y_height = int(random.uniform(-1, 3) + math.sin(x*0.2) + math.cos(z*0.2))
            y_offset = max(0, y_height) # Ensure ground level is at least 0

            b_type = 'grass'
            if x > 35: b_type = 'sand' # Larger arid zone
            if random.random() > 0.98: b_type = 'ore' # Sparser mining nodes

            Voxel(position=(x, y_offset, z), block_type=b_type)

            # Add water at a certain height level
            if y_offset < 0.5: # If terrain is low enough, add water
                Voxel(position=(x, 0, z), block_type='water')
'''
    game_code = game_code[:old_generate_func_start] + new_generate_imaland_code + game_code[old_generate_func_end:]
    print("Modified 'generate_imaland' function to include water and more varied terrain.")

# Note: math module is used in the new generate_imaland. We need to ensure it's imported.
# Since 'random' is already imported, we'll check and add 'math' if necessary.
if "import math" not in game_code:
    import_random_pos = game_code.find("import random")
    if import_random_pos != -1:
        game_code = game_code[:import_random_pos] + "import math\n" + game_code[import_random_pos:]
    else:
        # Fallback if random isn't found (shouldn't happen here)
        app_init_pos = game_code.find("app = Ursina()")
        game_code = game_code[:app_init_pos] + "import math\n" + game_code[app_init_pos:]


# Display the modified generate_imaland section for verification
generate_section_start = game_code.find("def generate_imaland():")
generate_section_end = game_code.find("\n# --- Game Loop (Survival Logic) ---", generate_section_start)
print('\n--- Modified generate_imaland function ---')
print(game_code[generate_section_start:generate_section_end])
```

### Refactoring the `update` function

```python
# Let's refactor the 'update' function to make it more modular.
# We'll extract the stamina logic and storm warning pulse into separate helper functions.

old_update_func_start = game_code.find("def update():")
old_update_func_end = game_code.find("# --- Initialization ---", old_update_func_start)

if old_update_func_start != -1 and old_update_func_end != -1:
    new_update_logic = '''
def handle_stamina_drain():
    if state['stamina'] > 0:
        state['stamina'] -= time.dt * state['heat_level']
        stamina_bar.scale_x = (state['stamina'] / 100) * 0.3

def pulse_storm_warning():
    storm_text.alpha = (sin(time.time() * 5) + 1) / 2

def spawn_rain_particles():
    if state['is_storming'] and random.random() > 0.8:
        create_rain()

def update():
    handle_stamina_drain()
    pulse_storm_warning()
    spawn_rain_particles()
'''
    game_code = game_code[:old_update_func_start] + new_update_logic + game_code[old_update_func_end:]
    print("Refactored 'update' function into helper functions.")

# Display the modified update section for verification
update_section_start = game_code.find("def update():")
# Find the start of the next section, which is "# --- Initialization ---"
update_section_end = game_code.find("\n# --- Initialization ---", update_section_start)
# Also find the helper functions we just added, which are before 'def update():'
helper_functions_start = game_code.rfind("def handle_stamina_drain():", 0, update_section_start)

if helper_functions_start != -1 and update_section_end != -1:
    print('\n--- Refactored update and helper functions ---')
    print(game_code[helper_functions_start:update_section_end])
else:
    print('\n--- Could not find refactored update section for display ---')

# Finally, let's write the modified code to a new file, as the original run attempt used a file.
# This allows for consistent execution if the user attempts to run it again.
with open('imaland_game_modified.py', 'w') as f:
    f.write(game_code)

print('\nModified game code saved to imaland_game_modified.py')
```

### Expansion: Caves, Civilization, Portals, and Map System

```python
# 1. Update BLOCKS with more variety for civilization and portals
new_blocks = """
    'planks': {'color': color.rgb(120, 80, 50), 'texture': 'white_cube'},
    'brick':  {'color': color.rgb(150, 50, 50), 'texture': 'white_cube'},
    'portal': {'color': color.rgb(0, 255, 100), 'texture': 'white_cube'}, # Glowing Portal
"""

blocks_start = game_code.find("BLOCKS = {")
blocks_end = game_code.find("}", blocks_start)
game_code = game_code[:blocks_end] + new_blocks + game_code[blocks_end:]

# 2. Redefine world generation to include civilization and caves
new_gen_logic = '''
def generate_structure(x, y, z):
    # Simple ruin/room generator
    for dx in range(3):
        for dz in range(3):
            for dy in range(3):
                if dy == 0 or (dx % 2 == 0 or dz % 2 == 0):
                    Voxel(position=(x+dx, y+dy, z+dz), block_type='brick')

def generate_imaland():
    for z in range(60):
        for x in range(60):
            # Terrain and Caves
            y_height = int(random.uniform(-1, 4) + math.sin(x*0.1) * 2)
            y_offset = max(0, y_height)

            # Sub-surface caves
            is_cave = random.random() > 0.92 and y_offset > 1

            if not is_cave:
                b_type = 'grass' if x < 40 else 'sand'
                Voxel(position=(x, y_offset, z), block_type=b_type)

                # Rare Civilization Outposts
                if random.random() > 0.995:
                    generate_structure(x, y_offset + 1, z)

                # Portals to New Worlds
                if random.random() > 0.999:
                    for py in range(3):
                        Voxel(position=(x, y_offset+1+py, z), block_type='portal')

            # Underwater logic / Caves
            if y_offset < 1:
                Voxel(position=(x, 0, z), block_type='water')
'''

gen_start = game_code.find("def generate_imaland():")
gen_end = game_code.find("def handle_stamina_drain():")
game_code = game_code[:gen_start] + new_gen_logic + game_code[gen_end:]

# 3. Add Mini-Map UI
map_ui_code = """
mini_map = Entity(parent=camera.ui, model='quad', scale=(0.15, 0.15),
                  position=(-0.7, 0.4), color=color.rgba(0,0,0,150))
map_player_dot = Entity(parent=mini_map, model='circle', scale=(0.05, 0.05), color=color.red)
"""

init_pos = game_code.find("player = FirstPersonController()")
game_code = game_code[:init_pos] + map_ui_code + game_code[init_pos:]

# 4. Update the Map in the loop
map_update = """    # Update Map Dot Position
    map_player_dot.x = (player.x / 60) - 0.5
    map_player_dot.y = (player.z / 60) - 0.5
"""

update_pos = game_code.find("spawn_rain_particles()")
insert_map_pos = game_code.find("\n", update_pos) + 1
game_code = game_code[:insert_map_pos] + map_update + game_code[insert_map_pos:]

with open('imaland_expanded.py', 'w') as f:
    f.write(game_code)

print("Game expanded with Caves, Civilization, Portals, and a Mini-Map!")
```

```python
import json

# 1. Add Tools and Appliances to BLOCKS
crafting_blocks = """
    'workbench': {'color': color.rgb(100, 80, 20), 'texture': 'white_cube'},
    'furnace':   {'color': color.rgb(60, 60, 60),  'texture': 'white_cube'},
"""

blocks_end = game_code.find("}", game_code.find("BLOCKS = {"))
game_code = game_code[:blocks_end] + crafting_blocks + game_code[blocks_end:]

# 2. Add Auto-Save and Teleportation Logic
advanced_logic = """
def save_game():
    data = {
        'position': (player.x, player.y, player.z),
        'stamina': state['stamina']
    }
    with open('save_data.json', 'w') as f:
        json.dump(data, f)
    print("Game Auto-Saved!")

save_timer = 0

def check_teleportation():
    # If player is near a portal block, reset to start (teleport simulation)
    # In a local environment, this could trigger a scene change
    if state.get('teleporting'): return

    for v in scene.entities:
        if hasattr(v, 'block_type') and v.block_type == 'portal':
            dist = (player.position - v.position).length()
            if dist < 1.5:
                player.position = (0, 10, 0)
                print("Teleported to a New World!")
                break
"""

# Insert logic before the update function
update_pos = game_code.find("def update():")
game_code = game_code[:update_pos] + advanced_logic + game_code[update_pos:]

# 3. Inject into the update loop
update_injection = """
    global save_timer
    save_timer += time.dt
    if save_timer >= 30: # Save every 30 seconds
        save_game()
        save_timer = 0

    check_teleportation()
"""

# Find the end of the update function signature and insert
loop_insert_pos = game_code.find("spawn_rain_particles()")
loop_insert_pos = game_code.find("\n", loop_insert_pos) + 1
game_code = game_code[:loop_insert_pos] + update_injection + game_code[loop_insert_pos:]

with open('imaland_final.py', 'w') as f:
    f.write(game_code)

print("Final version saved: imaland_final.py")
```
